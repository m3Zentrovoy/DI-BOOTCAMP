```python
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill

# new workbook
wb = Workbook()
ws = wb.active
ws.title = "Grades"

# names and grades
data = {
    "Joe": {"math": 65, "science": 78, "english": 98, "gym": 89},
    "Bill": {"math": 55, "science": 72, "english": 87, "gym": 95},
    "Tim": {"math": 100, "science": 45, "english": 75, "gym": 92},
    "Sally": {"math": 30, "science": 25, "english": 45, "gym": 100},
    "Jane": {"math": 100, "science": 100, "english": 100, "gym": 60}
}

# Add header
headers = ["Student", "Math", "Science", "English", "Gym"]
ws.append(headers)

# Make the header bold and blue
header_font = Font(bold=True, color="FFFFFF")
header_fill = PatternFill(start_color="4F81BD", end_color="4F81BD", fill_type="solid")
for cell in ws[1]:
    cell.font = header_font
    cell.fill = header_fill

# Add student data to the sheet
for student, grades in data.items():
    ws.append([
        student,
        grades["math"],
        grades["science"],
        grades["english"],
        grades["gym"]
    ])

# Write average formula for each subject in the last row
row_count = len(data) + 2
ws[f"A{row_count}"] = "Average"
ws[f"B{row_count}"] = f"=AVERAGE(B2:B{row_count-1})"
ws[f"C{row_count}"] = f"=AVERAGE(C2:C{row_count-1})"
ws[f"D{row_count}"] = f"=AVERAGE(D2:D{row_count-1})"
ws[f"E{row_count}"] = f"=AVERAGE(E2:E{row_count-1})"


ws[f"A{row_count}"].font = Font(bold=True)
wb.save("Grades.xlsx")
```
