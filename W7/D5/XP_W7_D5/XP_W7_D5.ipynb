**🌟 Exercise 3 : Extracting Headers from Wikipedia’s Main Page**

---


```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

url = 'https://en.wikipedia.org/wiki/Main_Page'

response = urlopen(url)
html_content = response.read()

soup = BeautifulSoup(html_content, 'html.parser')
soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6'])
```

**🌟 Exercise 4 : Checking for Page Title**

---

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

url = 'https://en.wikipedia.org/wiki/Main_Page'

response = urlopen(url)
html_content = response.read()

soup = BeautifulSoup(html_content, 'html.parser')
if soup.title:
    print("The page has a title:", soup.title.get_text())
else:
    print("No title tag found on the page.")
```

**🌟 Exercise 5 : Analyzing US-CERT Security Alerts**

---

```python
from urllib.request import urlopen
from bs4 import BeautifulSoup

url = 'https://www.cisa.gov/news-events/cybersecurity-advisories?search_api_fulltext=US-CERT&sort_by=field_release_date&url=&f%5B0%5D=advisory_type%3A93&f%5B1%5D=release_date_year%3A2023'
response = urlopen(url)
html = response.read()
soup = BeautifulSoup(html, 'html.parser')

# Найдём все карточки с алертами
alerts = soup.find_all('h3', class_='c-teaser__title')
print(f"Found {len(alerts)} US-CERT security alerts for 2023.")
```

**🌟 Exercise 6 : Scraping Movie Details**

---


```python
import requests

url = "https://www.imdb.com/list/ls091294718/"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
html = response.text
```

```python
from bs4 import BeautifulSoup
import json

soup = BeautifulSoup(html, "html.parser")
script_tag = soup.find('script', type="application/ld+json")
```

```python
if script_tag:
    data = json.loads(script_tag.string)
    movies = data.get('itemListElement', [])
else:
    print("JSON-LD not found.")
```

```python
import random

if len(movies) >= 10:
    random.shuffle(movies)
    chosen = movies[:10]
    for i, m in enumerate(chosen, 1):
        item = m['item']
        name = item['name']
        description = item['description']
        url = item['url']
        print(f"{i}. {name} URL: {url}")
else:
    print(f"found: {len(movies)}")
```
