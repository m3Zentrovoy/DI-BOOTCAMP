```python

#🌟 Exercise 2: Descriptive Statistics


data = [12, 15, 13, 12, 18, 20, 22, 21]
import scipy.stats as stats

mean_scipy = stats.tmean(data)
median_scipy = np.median(data)  # SciPy does not have a separate median function
variance_scipy = stats.tvar(data)
std_dev_scipy = stats.tstd(data)

print(f"Mean (SciPy): {mean_scipy}")
print(f"Median (SciPy): {median_scipy}")
print(f"Variance (SciPy): {variance_scipy}")
print(f"Standard Deviation (SciPy): {std_dev_scipy}")
```

```python
#🌟 Exercise 4: T-Test Application

from scipy.stats import ttest_ind

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Perform T-test
t_stat, p_val = ttest_ind(data1, data2)
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_val}")
```

```python
#🌟 Exercise 5: Linear Regression Analysis

import numpy as np
from scipy import stats

# Example data
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Performing linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

# The slope and intercept of the regression line
print("Slope:", slope)
print("Intercept:", intercept)
# %%
import matplotlib.pyplot as plt

# The R-squared value
print("R-squared:", r_value**2)

# P-value for the slope
print("P-value:", p_value)

# Standard error of the estimate
print("Standard error:", std_err)

# Make predictions using the regression line
predicted_house_prices = intercept + slope * house_sizes

# Plotting the original data and the regression line
plt.scatter(house_sizes, house_prices, label='Original Data')
plt.plot(house_sizes, predicted_house_prices, color='red', label='Regression Line')
plt.xlabel('House Size')
plt.ylabel('House Price')
plt.title('Linear Regression of House Price vs. House Size')
plt.legend()
plt.show()

# %%
# Predict the price for a 90 sq meter house
house_size_to_predict = 90
predicted_price = intercept + slope * house_size_to_predict

print(f"Predicted price for a {house_size_to_predict} sq meter house: {predicted_price:.2f}")
```

**The slope is the increase in price depending on the size of the house (square meters).**

```python
#🌟 Exercise 6: Understanding ANOVA


from scipy import stats

# Sample data from three different groups
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Performing one-way ANOVA
f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("F-value:", f_value)
print("P-value:", p_value)
```

* Yes — the fertilizers have significantly different effects on plant growth.


* The P-value is very small (less than 0.05), which means the differences in plant growth across the fertilizer groups are statistically significant.

```python
#🌟 Exercise 7: Probability Distributions (Optional)

# Example: Binomial Distribution
n, p = 10, 0.5  # number of trials, probability of each trial
binom_dist = stats.binom(n, p)
print("PMF 5 heads in 10 coin flips is:", binom_dist.pmf(5))



# Plotting Binomial Distribution
x_binom = np.arange(0, n+1)
pmf_binom = binom_dist.pmf(x_binom)
plt.figure(figsize=(6, 4))
plt.bar(x_binom, pmf_binom)
plt.title('Binomial Distribution\n$n=10$, $p=0.5$')
plt.xlabel('Number of Successes')
plt.ylabel('PMF')
plt.show()
```

```python
#🌟 Exercise 8: Correlation Coefficients (Optional)

import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Твои данные
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

# Pearson Correlation
pearson_corr, _ = pearsonr(data['age'], data['income'])
print(f'Pearson Correlation Coefficient: {pearson_corr:.2f}')

# Spearman Correlation
spearman_corr, _ = spearmanr(data['age'], data['income'])
print(f'Spearman Rank Correlation: {spearman_corr:.2f}')
```

###1.00 indicates a perfect positive linear relationship,
