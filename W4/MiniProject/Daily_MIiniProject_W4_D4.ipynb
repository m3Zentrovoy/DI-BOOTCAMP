  ---

  * Calculation of central tendency measures (mean, median, mode).

---

```python
import pandas as pd

mean = df.mean() # mean
median = df.median() # median
mode = df.mode().iloc[0] # mode


# сontain to one DataFrame
summary_df = pd.DataFrame({
    'Mean': mean,
    'Median': median,
    'Mode': mode
})

summary_df
```

 ---

  * Analysis of variability (range, variance, standard deviation).
Evaluation of distribution shapes through skewness and kurtosis.

---

```python

# define metrics
summary_df['Range'] = df.max() - df.min() # range
summary_df['Variance'] = df.var() # variance
summary_df['Standard Deviation'] = df.std() # standard deviation

summary_df
```

 ---

  * Evaluation of distribution shapes through skewness and kurtosis.

  ---

```python
from scipy.stats import skew, kurtosis
# define metrics
summary_df['Skewness'] = df.skew() # skewness
summary_df['Kurtosis'] = df.kurtosis() # kurtosis

summary_df

```

**Evaluation of distribution shapes through skewness and kurtosis.**
* Skewness:

  * 0: Right-skewed (tail extends to the right)

  *  < 0: Left-skewed (tail extends to the left)
  * ≈ 0: Symmetric distribution
* Kurtosis:

  * 3: Leptokurtic (high peak, heavy tails)

  * ≈ 3: Mesokurtic (normal distribution)
  * < 3: Platykurtic (flat distribution, light tails)

* Perform hypothesis testing for statistical significance between groups (e.g., different price ranges).

---

```python
from scipy import stats

# Sample data from three different groups
group0 = df[df['price_range'] == 0]['ram']
group1 = df[df['price_range'] == 1]['ram']
group2 = df[df['price_range'] == 2]['ram']
group3 = df[df['price_range'] == 3]['ram']

# Performing one-way ANOVA
f_value, p_value = stats.f_oneway(group0, group1, group2, group3)
print("F-value:", f_value)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between the groups.")
else:
    print("There is no significant difference between the groups.")
```

---

Different price range levels are strongly related to the amount of RAM. That is: **the higher** the price segment, **the more** RAM the device has.

* Investigate feature-target correlations using SciPy.

----

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

# Select features and target
features = ["battery_power", "ram", "px_height", "px_width", "int_memory", "talk_time"]
target = "price_range"

# Initialize empty dictionary to store correlation results
correlations = {
    "Pearson": [],
    "Spearman": []
}

# Compute correlation for each feature
for feature in features:
    pearson_corr, _ = pearsonr(df[feature], df[target])
    spearman_corr, _ = spearmanr(df[feature], df[target])

    correlations["Pearson"].append(round(pearson_corr, 6))
    correlations["Spearman"].append(round(spearman_corr, 6))

# Create a DataFrame with features as index
corr_df = pd.DataFrame(correlations, index=features)

# Display result
print("Correlation Results:")
print(corr_df)
```

* f high positive Pearson correlation (close to +1) → Strong linear relationship.
* If high positive Spearman correlation (close to +1) → Strong monotonic relationship.
* A low correlation (close to 0) → Weak or no relationship.

---
Apply advanced SciPy statistical functions for deeper insights.

---

```python
from scipy.stats import normaltest

# Select numerical features (excluding target)
features = ["battery_power", "ram", "px_height", "px_width", "int_memory", "talk_time"]

# Store results
results = []

for feature in features:
    stat, p = normaltest(df[feature])
    results.append({
        'Feature': feature,
        'Statistic': stat,
        'P-value': p,
        'Normal?': 'Yes' if p > 0.05 else 'No'
    })

# Convert results to DataFrame
normality_df = pd.DataFrame(results)
print(normality_df)
```

---

Most features, such as **ram, px_height,** and **px_width, do not follow a normal distribution** (p-value < 0.05), indicating skewed or non-Gaussian distributions.
Some features like **battery_power, int_memory**, and **talk_time may be normally distributed**, based on their higher p-values.

**4. Data Visualization with Matplotlib:**

* Produce histograms, scatter plots, and box plots for data distribution and relationship insights.
* Employ heatmaps for correlation visualization.
* Ensure clarity in plots with appropriate titles, labels, and axis information.

---

```python
import matplotlib.pyplot as plt
import seaborn as sns


# Selecting important features
features = ["battery_power", "ram", "px_height", "px_width", "int_memory", "talk_time"]
target = "price_range"


# Show features distributions
plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 3, i+1)
    sns.histplot(df[feature], kde=True, color='orchid', bins=30)
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Frequency')
plt.tight_layout() # adjusts the spacing between subplots to prevent overlapping and ensure they fit nicely within the figure.
plt.show()
```

```python

# feature-target relationships

plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 3, i+1)
    sns.scatterplot(x=df[feature], y=df[target], color='pink', alpha=0.5)
    plt.title(f'{feature} vs. {target}')
    plt.xlabel(feature)
    plt.ylabel(target)
plt.tight_layout()
plt.show()
```

```python

# Identify outliers
plt.figure(figsize=(12, 8))
for i, feature in enumerate(features):
    plt.subplot(2, 3, i+1)
    sns.boxplot(x=df[target], y=df[feature], hue = df[target],  palette="coolwarm")
    plt.title(f'Boxplot of {feature}')
    plt.xlabel(feature)
    plt.legend(title=target, loc='upper left')

plt.tight_layout()
plt.show()
```

*  **Internal memory** and talk time do not show a clear trend, indicating they may not be strong price predictors.

**Conclusion:**

* Manufacturers looking to categorize phones into price ranges should prioritize **RAM** size.
